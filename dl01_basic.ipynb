PyTorch에서 GPU가속 사용 가능 확인

```python
import torch
import torch.nn as nn

# 1. 실행 환경에 맞는 장치(Device) 동적 할당 (표준 패턴)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("현재 사용 장치:", device)

# 2. 텐서 생성 시 처음부터 지정된 장치로 할당 (메모리 이동 오버헤드 방지)
x = torch.ones((2, 3), device=device)

# 3. 기존 텐서를 지정 장치로 이동
# Note: 텐서의 .to()는 원본을 변경하지 않고 '새로운 텐서'를 반환합니다.
y = torch.tensor([[1.0, 2.0], [3.0, 4.0]])
y = y.to(device)

# 4. 신경망 모델을 지정 장치로 이동
# Note: 모델의 .to()는 모델 내부 파라미터를 'In-place(제자리)'로 이동시킵니다.
model = nn.Linear(2, 2).to(device)

# 5. 모델 연산 실행
output = model(y)
print("출력 결과 장치:", output.device)
```

## 1. 1 데이터로부터 직접 생성(From Data)

```python
a = torch.tensor([[1, 2], [3, 4]])
print(a)
print(type(a))  # <class 'torch.Tensor'>
print(a.dtype)  # torch.int64



```

```python

# python 리스트로부터 생성
list_data = [[1, 2, 3], [4, 5,6]]
print(f'list_data: {list_data}')
print(type(list_data))      # <class 'list'>
tensor_from_list = torch.tensor(list_data, dtype=torch.float32)
print(f'List -> Tesnor : {tensor_from_list}')
print(type(tensor_from_list))   # <class 'torch.Tensor'>
print(tensor_from_list.dtype)
print(tensor_from_list.device)
```

```python
# numpy 배열로 부터 생성
np_array = np.array([10, 20, 30])
# np_array = [[1, 2, 3], [4, 5,6]]    # TypeError: expected np.ndarray (got list), 리스트는 지원하지 않는다.

# from_numpy()으로 tensor()을 생성하면 메모리를 공유한다.
tensor_from_np = torch.from_numpy(np_array)     # from_numpy : 메모리 공유
print(f'tensor_from_np : {tensor_from_np}')

tensor_from_np[0] = 999
print(f'np_array : {np_array}')     # 999
print(f'tensor_from_np : {tensor_from_np}')     # 999

np_array[0] = 500
print(f'np_array : {np_array}')     # 500
print(f'tensor_from_np : {tensor_from_np}')     # 500


```

## 1.2 특정 형태(Shape)와 값으로 초기와(Structured Initialization)

```python
shape = (2, 3)
print(f'zeros : {torch.zeros(shape)}')
print(f'ones : {torch.ones(shape)}')
print(f'empty : {torch.empty(shape)}')
print(f'full : {torch.full(shape, 3)}')

# 범위 기반 생성
print(f'arange : {torch.arange(10)}')       # end
print(f'arange : {torch.arange(0, 10)}')    # start, end
print(f'arange : {torch.arange(0, 10, 2)}') # start, end, step

print(f'linspace : {torch.linspace(1, 10, 5)}') # shape = (2, 3)
print(f'zeros : {torch.zeros(shape)}')
print(f'ones : {torch.ones(shape)}')
print(f'empty : {torch.empty(shape)}')
print(f'full : {torch.full(shape, 3)}')

# 범위 기반 생성
print(f'arange : {torch.arange(10)}')       # end
print(f'arange : {torch.arange(0, 10)}')    # start, end
print(f'arange : {torch.arange(0, 10, 2)}') # start, end, step

print(f'linspace : {torch.linspace(1, 10, 5)}') # start, end, step(1부터 10까지 5등분), linspace : tensor([ 1.0000,  3.2500,  5.5000,  7.7500, 10.0000])
# np.linspase(1, 10 5, endpoint = False)




```

## 1.3 무작위 난수로 초기화(Random, Initialization)

```python
# 난수 생성 시 재현성을 위한 시드(seed) 고정
torch.manual_seed(42)

print(f'rand : {torch.rand((2, 2))}')

print(f'rand : {torch.rand(2, 2)}')      # *size

print(f'randn : {torch.randn(2, 2)}')    # 정규분포를 따르는 난수 발생

print(f'randint : {torch.randint(10, (2, 2))}') # (10, (2, 2)) : (high, size=())
print(f'randint : {torch.randint(1, 10, (2, 2))}')      # (1, 10, (2, 2)) : (low, high, size())
print(f'randint : {torch.randint(low=1, high=10, size=(2, 2))}') 

# 0부터 n-1까지만의 정수를 무작위로 섞인 1차 텐서 생성
print(f'randperm : {torch.randperm(10)}') # randperm : tensor([7, 2, 6, 8, 5, 9, 0, 3, 4, 1])


        

```

## 1.4 기존 텐서의 속성을 복사하여 초기화(_like 계열)

```python
# x = torch.tensor([[1, 2, 3], [4, 5, 6]])

x = torch.tensor([[1, 2, 3], [4, 5, 6]], dtype=torch.float32)
print(f'x : {x}')

print(f'zeros_like: {torch.zeros_like(x)}')     # 모양과 데이터 타입고 복사됨

# shape와 dtype을 그대로 복사해 온다
x_zeros = torch.zeros_like(x)
print(f'zeros_like : {x_zeros}, dtype : {x_zeros.dtype}')

x_ones = torch.ones_like(x)
print(f'ones_like : {x_ones}, dtype : {x_ones.dtype}')

x_rand = torch.rand_like(x)     # rand_like() : 실수 값만 가질수 있고 정수값을 가질 수 없어 NotImplementedError: "check_uniform_bounds" not implemented for 'Long' 오류 발생
print(f'rand_like : {x_rand}, dtype : {x_rand.dtype}')

# randint_like()은 x의 타입이 float32이므로 그 타입을 그대로 따라서 사용이 된다.(주의)
x_randint = torch.randint_like(x, 10)   
print(f'randint_like : {x_randint}, dtype{x_randint.dtype}')

```
